- longitude = decimal degrees [-360,360]
- depth = depth of the event (km)
- mag =  magnitude of event
- id = event identifier
- place = where the event took place
- type = type of event

# 6
Without creating any new variables, plot a histogram of the eartquake's magnitudes (`eqk` data). Add a title and change the color of your graph.

```python
# Plotting histogram of earthquake magnitudes
eqk['mag'].plot.hist(title='Earthquake Magnitudes',
                     bins=50,
                     xlabel='Magnitude',
                     ylabel='Counts',
                     color='pink',
                     edgecolor='black');
```

# 7

a. Create a new variable `top20` with the 20 earthquakes with largest magnitude. HINT: check the method [`pandas.Series.nlargest`](https://pandas.pydata.org/docs/reference/api/pandas.Series.nlargest.html). Your answer should contain only the `id` and `mag` columns. 

```python
# 20 earthquakes with largest magnitude
top20 = eqk['mag'].nlargest(n=20)
```

b. Create a bar plot showing the magnitude of the top 20 earthquakes. Update the title, x-axis label, y-axis label, and color of the bars. 

```python
# Graphing magnitude top 20 earthquakes
top20.plot.bar(top20.index,
               title='Top 20 Earthquake Magnitudes', 
               xlabel='Earthquake ID', 
               ylabel='Magnitude', 
               color='pink', 
               edgecolor='black');
```

<!-- BEGIN QUESTION -->

c. What is the role of the `top20` index in the graph? What would have happened if we had not updated the index?

> The `top20` index in the graph corresponds to the x-coordinates of the bars. If we had not updated the index, then the indices would have mostly been very large numbers, since we filtered for the 20 largest magnitude earthquakes from 120,000+ rows. This is too big of a range of x-coordinates to plot on a bar plot, and it would make each bar too small for us to see. Therefore, it was best to set the indices to a qualitative values, like `id`.

# 8 
We are interested in the place, magnitude, and depth of the top 20 earthquakes with largest magnitude. Select this data from the `eqk` data frame. Store your answer in the `top20_filtered` variable. 

HINT: This is a "select rows and columns simultaneously" from `eqk` exercise. To select the rows you could use the index of `top20`. Make sure your columns are in the following order to pass the test: `place`, `mag`, `depth`. 

```python
# Filtering df, selecting certain columns, and the 20 largest magnitude earthquakes
top20_filtered = eqk[['place', 'mag', 'depth']].loc[eqk['mag'].nlargest(n=20).index]
```

# 9 

Without creating any new variables, visualize the locations of eartquakes with magnitude greter than 5 by creating a scatter plot of their latitude and longitude. Make the following adjustments to the basic plot:
- update the title
- color the points by magnitude (HINT: [check the `c` parameter here](https://pandas.pydata.org/pandas-docs/stable/reference/api/pandas.DataFrame.plot.scatter.html))
- adjust the `alpha` (transparency) argument
- adjust the `colormap` argument to maket it look prettier. [Click for ideas of sequential colormaps.](https://matplotlib.org/stable/users/explain/colors/colormaps.html)


```python
# Scatter plot of earthquakes with mags greater than 5
eqk[eqk['mag']>5].plot.scatter(
    x = 'latitude',
    y = 'longitude',
    title = 'Earthquakes with Magnitude Greater Than 5',
    c = 'mag',             # coloring each point
    alpha = 0.8,           # adjusting transparency
    colormap='RdPu_r'      # choosing a colormap
);
```
